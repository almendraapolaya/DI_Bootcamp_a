Daily Challenge: End-to-End Web Scraping In Python
===

###What you will create?
A Python script to scrape the GitHub Topics webpage.
Utilizing BeautifulSoup to parse the saved HTML content of the GitHub Topics page.
A dictionary into a pandas DataFrame for better data manipulation and analysis.

##Step 1: Fetch and Save the Webpage

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = 'https://github.com/topics'
response = requests.get(url)

print(f"Status Code: {response.status_code}")

print(f"First 100 chars: {response.text[:100]}")

with open('webpage.html', 'w', encoding='utf-8') as f:
    f.write(response.text)
```

##Step 2: Parse and Extract Data

```python
with open('webpage.html', 'r', encoding='utf-8') as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, 'html.parser')

topic_title_tags = soup.find_all('p', class_='f3 lh-condensed mb-0 mt-1 Link--primary')
topic_desc_tags = soup.find_all('p', class_='f5 color-fg-muted mb-0 mt-1')

titles = [tag.get_text().strip() for tag in topic_title_tags]
descriptions = [tag.get_text().strip() for tag in topic_desc_tags]

print(f"Titles found: {len(titles)}")
print(f"Descriptions found: {len(descriptions)}")
print(f"Sample Title: {titles[0] if titles else 'None'}")
```

##Step 3: Structure and Analyze with Pandas

```python
topics_dict = {
    'title': titles,
    'description': descriptions
}

df = pd.DataFrame(topics_dict)

print("\n--- GitHub Topics DataFrame ---")
print(df.head())
```
