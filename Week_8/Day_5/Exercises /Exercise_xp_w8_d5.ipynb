## Exercise 2 : Scraping robots.txt from Wikipedia


```python
import urllib.request
```

```python
url = "https://en.wikipedia.org/robots.txt"

request = urllib.request.Request(
    url,
    headers={'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'}
)

try:
    with urllib.request.urlopen(request) as response:
        robots_content = response.read().decode('utf-8')

    print("--- Wikipedia robots.txt Content ---")
    print(robots_content[:500])
    print("\n... [Content truncated for brevity] ...")

except Exception as e:
    print(f"An error occurred: {e}")
```

## Exercise 3 : Extracting Headers from Wikipedia’s Main Page

```python
import urllib.request
from bs4 import BeautifulSoup
```

```python
url = "https://en.wikipedia.org/wiki/Main_Page"

request = urllib.request.Request(
    url,
    headers={'User-Agent': 'Mozilla/5.0'}
)

try:
    with urllib.request.urlopen(request) as response:
        html_content = response.read()

    soup = BeautifulSoup(html_content, 'html.parser')

    headers = soup.find_all(['h1', 'h2', 'h3', 'h4', 'h5', 'h6'])

    print(f"Total headers found: {len(headers)}\n")

    for header in headers:
        print(f"[{header.name}] {header.get_text().strip()}")

except Exception as e:
    print(f"An error occurred: {e}")
```

```python
import urllib.request
from bs4 import BeautifulSoup
```

## Exercise 5 : Analyzing US-CERT Security Alerts

```python
import urllib.request
from bs4 import BeautifulSoup
from datetime import datetime
```

```python

url = "https://www.cisa.gov/news-events/cybersecurity-advisories?f%5B0%5D=advisory_type%3A93"

request = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})

try:
    with urllib.request.urlopen(request) as response:
        html_content = response.read()

    soup = BeautifulSoup(html_content, 'html.parser')

    current_year = str(datetime.now().year)

    alerts = soup.find_all('div', class_='views-row')

    count = 0
    for alert in alerts:
        if current_year in alert.get_text():
            count += 1

    print(f"--- CISA/US-CERT Security Alerts ---")
    print(f"Number of alerts found for {current_year} on the first page: {count}")

except Exception as e:
    print(f"An error occurred: {e}")
```
