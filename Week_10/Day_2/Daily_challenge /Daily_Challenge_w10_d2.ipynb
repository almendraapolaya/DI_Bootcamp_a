**Task 1 — Calculate Required Sample Size**


```python
import math
import statsmodels.stats.api as sms

# Given parameters
alpha = 0.05       # 5% Significance level
power = 0.80       # 80% Statistical power
effect_size = 0.2  # Expected Cohen's h effect size

# Initialize power analysis calculator for two independent samples
calculator = sms.NormalIndPower()

# Calculate required sample size per group
required_n = calculator.solve_power(
    effect_size=effect_size,
    power=power,
    alpha=alpha,
    ratio=1.0,          # Equal split (1:1 ratio)
    alternative='two-sided'
)

sample_per_group = math.ceil(required_n)

print("=== TASK 1: BASELINE SAMPLE SIZE CALCULATION ===")
print(f"Effect Size (Cohen's h) : {effect_size}")
print(f"Significance Level (α)  : {alpha}")
print(f"Statistical Power (1-β) : {power}")
print("-" * 45)
print(f"Required Sample Size per Group: {sample_per_group:,} visitors")
print(f"Total Visitors Needed (A + B) : {sample_per_group * 2:,} visitors")
```

**Task 2 — Analyze the Impact of Varying Effect Sizes**

```python
import math
import statsmodels.stats.api as sms
import pandas as pd

# Define list of effect sizes to analyze
effect_sizes = [0.1, 0.2, 0.3, 0.4]
alpha = 0.05
power = 0.80

calculator = sms.NormalIndPower()

results = []

for es in effect_sizes:
    n = calculator.solve_power(
        effect_size=es,
        power=power,
        alpha=alpha,
        ratio=1.0,
        alternative='two-sided'
    )
    n_ceil = math.ceil(n)
    results.append({
        "Effect Size (Cohen's h)": es,
        "Sample Size per Group": f"{n_ceil:,}",
        "Total Sample Size": f"{n_ceil * 2:,}"
    })

# Display as a clean pandas DataFrame
df_results = pd.DataFrame(results)
print("=== TASK 2: EFFECT SIZE IMPACT ANALYSIS ===")
print(df_results.to_string(index=False))
```

**Task 2 & 3 Explanations**

### Results & Analysis

| Effect Size (Cohen's $h$) | Sample Size per Group | Total Sample Size |
| :--- | :--- | :--- |
| **0.1** (Very Small) | **1,571** | **3,142** |
| **0.2** (Small) | **393** | **786** |
| **0.3** (Medium) | **175** | **350** |
| **0.4** (Large) | **99** | **198** |

#### Why does the sample size drop so rapidly?
The required sample size is **inversely proportional to the square of the effect size** ($\text{Sample Size} \propto \frac{1}{\text{Effect Size}^2}$).
* A **small effect size (0.1)** produces a faint signal, requiring **1,571 visitors per group** to separate real impact from background random noise.
* A **large effect size (0.4)** produces a loud, unmistakable signal, allowing us to prove statistical significance with only **99 visitors per group**.

### Explaining to the "Sweet Bytes" Bakery Team

Imagine you’re baking two batches of cookies to see if a new recipe is sweeter:

* **Slight Recipe Tweak (Small Effect Size = 0.1):** The new cookie is only slightly sweeter than the old one. Because the taste difference is subtle, you need a **huge crowd of 3,142 taste testers** to be 80% sure that people genuinely prefer the new recipe rather than just guessing randomly.
* **Double Chocolate Overload (Large Effect Size = 0.4):** The new cookie is dramatically richer and sweeter. The difference is so obvious that you only need **198 taste testers** to confidently confirm the winner!

#### Why Balancing Effect Size & Sample Size Matters:
1. **Saves Bakery Resources:** Testing takes time and online traffic. Knowing your required sample size prevents you from running tests for months unnecessarily.
2. **Avoids False Promises:** If we expect a massive boost (+4% conversion) but the actual improvement is tiny (+0.2%), stopping the test too early with a small sample size might cause us to miss a real improvement (**Type II error**) or declare a false winner (**Type I error**).
3. **Confident Decisions:** Proper power planning guarantees that when we roll out the new checkout process, "Sweet Bytes" will see a real, measurable increase in sales!
