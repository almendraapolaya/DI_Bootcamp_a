### Exercise 1: Defining the Problem and Data Collection for Loan Default Prediction

#### 1. Problem Statement
The objective of this project is to develop a binary classification model to predict the binary target variable `default_status` (where `1` indicates a borrower will fail to meet loan repayment obligations within a specified timeframe, and `0` indicates timely repayment) to support automated credit underwriting and risk-based pricing decisions. By accurately estimating individual default probabilities before loan origination, the financial institution can optimize approval thresholds, reduce non-performing loans, and maintain portfolio profitability while ensuring fair access to credit.

#### 2. Required Data Types & Justifications
* **Applicant Demographics & Income:** This category includes applicant age, employment status, annual income, job tenure, and debt-to-income (DTI) ratio.
  * *Justification:* Higher net income and lower debt-to-income ratios indicate greater financial capacity to service additional debt without triggering financial distress.
* **Loan Application Attributes:** This category includes the requested loan amount, loan term length, interest rate, and specified loan purpose.
  * *Justification:* Larger loan amounts relative to income increase monthly debt obligations, directly impacting the borrower's likelihood of default over time.
* **Historical Credit Performance:** This category includes credit bureau scores, total active credit lines, revolving credit utilization, and past delinquency or bankruptcy history.
  * *Justification:* Past repayment behaviors serve as strong indicators of future financial discipline and overall credit risk.
* **Transaction & Behavioral Data:** This category includes average monthly checking account balances, deposit consistency, overdraft frequency, and relationship tenure with the institution.
  * *Justification:* Real-time cash flow patterns capture immediate financial stability that static quarterly credit bureau scores might miss.

#### 3. Data Sources & Integration Strategy
* **Internal Core Banking Systems:** Historical customer profiles, deposit transaction records, and previous loan performance data will be extracted directly from the financial institution’s internal databases using SQL-based ETL pipelines.
* **Credit Reporting Bureaus:** Official credit scores and credit history details will be obtained by querying external credit bureau APIs (such as Experian, Equifax, or TransUnion) in real time during the application submission process.
* **Alternative Data & Digital Verification Providers:** Employment status and income details will be automatically verified by integrating third-party automated verification APIs (such as Plaid or Truework) directly into the online loan application workflow.

#### 4. Risks, Governance, and Regulatory Constraints
Developing a credit risk model involves strict regulatory frameworks, privacy mandates, and algorithmic bias risks. Regulations such as the Fair Credit Reporting Act (FCRA) and Equal Credit Opportunity Act (ECOA) strictly prohibit the inclusion of protected attributes (such as race, gender, or marital status) to prevent discriminatory lending practices. Additionally, data collection must comply with data privacy laws to ensure proper consent and secure storage. A primary challenge is historical sampling bias, as training datasets only contain performance data for previously approved applicants, potentially skewing predictions for rejected profiles.
```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

# Placeholders for true labels and predicted probabilities:
y_true = [0, 1, 0, 1, 0, 0, 1, 0, 1, 0]
y_pred_proba = [0.05, 0.80, 0.10, 0.65, 0.20, 0.15, 0.70, 0.30, 0.85, 0.25]

# Set a decision threshold that reflects the precision–recall trade-off:
threshold = 0.5
y_pred = [1 if p >= threshold else 0 for p in y_pred_proba]

print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_true, y_pred_proba), 4))
print("PR-AUC (Average Precision):", round(average_precision_score(y_true, y_pred_proba), 4))
print("\nConfusion matrix:\n", confusion_matrix(y_true, y_pred))
print("\nClassification report:\n", classification_report(y_true, y_pred, zero_division=0))
```

### Exercise 5: Designing an Evaluation Strategy for Different Machine Learning Models

#### 1. Supervised Learning Evaluation Strategy (Random Forest Classifier)
To evaluate a Random Forest classification model, I will execute a validation plan using 5-Fold Stratified Cross-Validation on the training set to prevent data leakage while maintaining target class proportions across folds. Performance will be evaluated using a comprehensive suite of metrics including Accuracy, Precision, Recall, F1-score, ROC-AUC, and PR-AUC. Diagnostic plots such as the Receiver Operating Characteristic (ROC) curve and Precision-Recall curve will be evaluated to analyze performance across various decision thresholds. The primary evaluation challenge for supervised classification models is handling severe class imbalance, where metrics like accuracy become misleadingly high despite poor performance on the minority class.

#### 2. Unsupervised Learning Evaluation Strategy (K-Means Clustering)
To assess the effectiveness of a K-Means clustering model without ground-truth labels, structural cohesion and separation must be evaluated using intrinsic metrics like the Silhouette Score and the Davies-Bouldin Index. Diagnostic visualization tools, specifically the Elbow Method plot (tracking Within-Cluster Sum of Squares across various values of $K$), will be used to identify the optimal number of clusters where diminishing returns occur. Additionally, Silhouette plots will be constructed to visually verify that individual clusters exhibit uniform density without substantial negative coefficient values. A key evaluation challenge in unsupervised learning is the absence of objective target labels, which makes validating real-world cluster utility reliant on domain expertise rather than definitive mathematical ground truths.

#### 3. Reinforcement Learning Evaluation Strategy (Deep Q-Network Agent)
To measure the success of a Deep Q-Network (DQN) agent navigating an environment, performance will be tracked using the moving average of cumulative rewards per episode over continuous training iterations. Model stability and convergence will be evaluated by monitoring policy value functions, loss decay, and step completion efficiency per episode. To manage the exploration versus exploitation trade-off, an $\epsilon$-greedy schedule will decay the exploration rate over time as the agent transitions from random exploration to exploiting learned optimal paths. A major evaluation challenge in reinforcement learning is high environment stochasticity and sample inefficiency, which makes training stability unstable and performance difficult to reproduce consistently across random seeds.
