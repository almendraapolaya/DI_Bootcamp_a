#Daily Challenge: Precision Prompting for Control & Output Quality


**What you will create**

You’ll design, test, and refine a prompt that performs four different control tasks in a realistic content generation workflow. You’ll work through a full prompt development cycle:

- Crafting a prompt with tone, structure, format, and length control
- Evaluating the output for quality and accuracy
- Mitigating hallucination risks
- Enhancing clarity through quoting or paraphrasing techniques


**Scenario**

You work on the Learning & Communications Team at a global tech company. You’ve been asked to use generative AI to create employee-facing microlearning content for an internal newsletter.

Your task is to turn a paragraph from a recent internal policy update into a concise and clear knowledge snippet. The original policy text is overly formal, dense, and full of jargon.

**Step 1: Craft a Prompt That Controls Output Tone, Format, and Length**

```python
# Step 1: Initial precision prompt
policy_text = """Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."""

step1_prompt = """Act as an Internal Communications Specialist writing a microlearning snippet for our company newsletter.

Rewrite the following policy text into a friendly, clear, and easy-to-read snippet:

Input Text:
"{policy_text}"

Constraints:
- Tone: Friendly, clear, and approachable for all employees
- Format: Bullet points
- Style: Paraphrase the original text (do NOT quote it directly)
- Length: Strictly under 75 words total
- Content: Retain all core security rules accurately"""

print(step1_prompt)
```

**Step 2: Evaluate the Output**

```python
# Step 2: Evaluation criteria dictionary
evaluation_criteria = {
    "Relevance": "High - Retains the core message regarding secure remote access and protected devices.",
    "Clarity": "High - Simplifies dense legalistic phrasing into digestible employee guidance.",
    "Structure": "Met - Successfully breaks down instructions into bulleted points.",
    "Tone": "Met - Shifts from overly formal and punitive to friendly and supportive.",
    "Length": "Met - Stays under the strict 75-word limit.",
    "Factual Accuracy": "Met - Accurately conveys the VPN requirement and device protection rules without dropping critical details.",
}

for criterion, rating in evaluation_criteria.items():
    print(f"{criterion}: {rating}")
```

**Step 3: Detect and Mitigate Hallucinations**

```python
# Step 3: Revised prompt with anti-hallucination grounding
step3_grounded_prompt = """Act as an Internal Communications Specialist writing a microlearning snippet for our company newsletter.

Rewrite the following policy text into a friendly, clear, and easy-to-read snippet:

Input Text:
"Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."

STRICT GROUNDING & CONSTRAINTS:
1. Grounding: Rely ONLY on the information provided in the input text. Do NOT add new tools, recommendations, consequences, or technologies not explicitly mentioned (e.g., do not add 2FA, passwords, or IT helpdesk links).
2. Tone: Friendly, clear, and approachable.
3. Format: Bullet points.
4. Style: Paraphrase the original text (no direct quotes).
5. Length: Under 75 words total."""

print(step3_grounded_prompt)
```

**Step 4: Paraphrasing Deep Dive**

```python
# Step 4: Paraphrase variant for junior interns
step4_intern_prompt = """Act as a supportive mentor writing a quick security tips blurb for new junior interns.

Paraphrase the following policy text in plain language using short, simple phrases:

Input Text:
"Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."

Constraints:
- Audience: Junior interns (18-22 years old)
- Length/Structure: Maximum of 4 bullet points, short phrases
- Vocabulary: Zero corporate or legal jargon (avoid terms like 'proprietary data', 'endpoint protection', 'established via')
- Tone: Highly supportive, encouraging, and informative"""

print(step4_intern_prompt)
```

**Step 5: Quote Extraction Variant**

```python
# Step 5: Quote extraction prompt and analysis
step5_quote_prompt = """Extract the single most important sentence as a direct quote from the policy text below that captures the core security rule. Do not alter any words.

Input Text:
"Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications." """

quote_analysis = {
    "When_quoting_is_appropriate": "Quoting is best in formal compliance training, legal updates, executive memos, or official policy documentation where precision, strict adherence, and legal enforceability are required.",
    "When_quoting_poses_a_risk": "Quoting poses a risk when communicating with non-technical staff, as dense jargon and legalistic phrasing can confuse readers, reduce engagement, and lead to policy non-compliance due to lack of clarity.",
}

print("Quote Prompt:\n", step5_quote_prompt)
print("\nAnalysis:\n", quote_analysis)
```
