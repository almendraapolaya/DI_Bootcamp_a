## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
# TODO: write the full multi-part prompt
quiz_prompt = """Act as a middle school science educator. Based on the provided article about volcanic eruptions, create educational content for 7th-grade students (ages 11–13).

Please generate the following:

1. Summary:
- Write a 2-bullet-point summary of the main points using a friendly tone and simple vocabulary.

2. Quiz:
- Create 3 age-appropriate multiple-choice questions based on the article.
- Each question must feature 3 options total: 1 correct answer and 2 distractors.
- Clearly label the correct answer key at the end of each question.

Format Constraints:
- Structure the output directly as slide content (Slide 1: Summary, Slide 2–4: Questions 1–3) so it can be easily copied into Google Slides."""

quiz_prompt
```

```python
# TODO: explain three improvements over the vague prompt
improvements = [
    "Age-appropriate calibration: Specifying 7th grade / 11–13 year-olds ensures vocabulary and concept difficulty match middle schoolers, avoiding output geared toward kindergarteners or high schoolers.",
    "Structural & answer-key constraints: Explicitly defining 3 questions with 1 correct answer and 2 distractors guarantees a consistent test format rather than random open-ended or true/false questions.",
    "Ready-to-use slide formatting: Requesting Google Slides-ready layout saves time by eliminating the need to manually reformat raw text into presentation slides.",
]
improvements
```
